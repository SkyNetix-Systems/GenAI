# Math for Data Science — Simple Explanations with Examples
This notebook gives **easy-to-understand explanations** of all core maths topics used in Data Science.
Each topic has:
- Simple definition
- Real-world relevance
- Numerical example
- Python example

Let's begin 🚀

## Setup

```
import numpy as np
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
```

## 1 — Vectors
**Definition:** A vector is just a list of numbers — represents features of data.

**Where used:** Machine learning models treat each input row as a vector.

**Example:**
Height, weight, age → `[170, 72, 26]`

```
v = np.array([170, 72, 26])
v
```

## 2 — Dot Product
**Definition:** Multiply matching elements + add them.

**Where used:** Neural networks, similarity, projections.

**Example:**

```
v = np.array([1, 2, 3])
w = np.array([4, 5, 6])
np.dot(v, w)
```

## 3 — Norm (Vector Length)
**Definition:** Distance of vector from origin.

**Where used:** Distance between points (KNN), regularization.


```
np.linalg.norm(np.array([3,4]))  # 5
```

## 4 — Matrices
**Definition:** 2D array used to represent data or transformations.

**Where used:** Linear regression, PCA, neural networks.


```
A = np.array([[1,2],[3,4]])
B = np.array([[2,0],[1,3]])
A @ B
```

## 5 — Matrix Inverse
**Where used:** Solving linear equations (not used heavily in big ML but conceptually important).

```
np.linalg.inv(np.array([[3,1],[1,2]]))
```

## 6 — Determinant
**Definition:** Tells if matrix is invertible.

**Where used:** PCA, transformations.

```
np.linalg.det(np.array([[3,1],[1,2]]))
```

## 7 — Eigenvalues & Eigenvectors
**Where used:** PCA, dimensionality reduction.

**Intuition:** Directions in which data varies the most.

```
M = np.array([[2,1],[1,2]])
vals, vecs = np.linalg.eig(M)
vals, vecs
```

## 8 — Singular Value Decomposition (SVD)
**Where used:** PCA, recommendation systems, compression.


```
X = np.array([[3,1,1],[1,3,1]])
U, s, VT = np.linalg.svd(X)
U, s, VT
```

## 9 — Derivatives
**Definition:** Rate of change.

**Where used:** Training ML models using gradient descent.

**Example:** f(x) = x² → f'(x) = 2x

```
def f(x): return x**2

def dfdx(x, h=1e-6): return (f(x+h)-f(x-h))/(2*h)
[dfdx(i) for i in [-2,-1,0,1,2]]
```

## 10 — Gradient (Multivariable derivative)
**Where used:** Optimization in ML.

Example: f(x, y) = x² + y² → grad = [2x, 2y]

```
def grad(x, y): return np.array([2*x, 2*y])
grad(3,4)
```

## 11 — Gradient Descent (Optimization)
**Goal:** Minimize loss function.

**Where used:** EVERY ML model (neural networks, regression).


```
lr = 0.1
x = 10.0
for _ in range(10):
    grad = 2*x  # derivative of x^2
    x = x - lr*grad
x
```

## 12 — Probability Basics
**Used in:** Naive Bayes, statistics, uncertainty in ML.


```
np.random.seed(0)
vals = np.random.choice([0,1], size=10, p=[0.3,0.7])
vals
```

## 13 — Expected Value
**Average outcome in probability.**

```
data = np.array([0,1])
probs = np.array([0.3,0.7])
np.sum(data*probs)
```

## 14 — Variance & Standard Deviation
**Used in:** Risk, uncertainty, error measurement.

```
vals = np.array([10,12,14,16])
np.var(vals), np.std(vals)
```

## 15 — Covariance & Correlation
**Used in:** Feature relationship, PCA.
Correlation = how strongly two variables move together.

```
x = np.random.randn(100)
y = 2*x + np.random.randn(100)*0.5
np.corrcoef(x,y)
```

## 16 — PCA (Dimensionality Reduction)
**Used in:** Feature reduction, noise removal.

PCA finds direction of maximum variance.

```
X = np.vstack([x,y]).T
X_centered = X - X.mean(axis=0)
U, s, VT = np.linalg.svd(X_centered)
VT[0]  # first principal component
```
