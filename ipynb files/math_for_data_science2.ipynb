# **1. Descriptive Statistics**

Descriptive statistics help us **summarize data**.
Let's say we have marks: `[12, 15, 14, 10, 18]`

## **1.1 Mean (Average)**
**Meaning:** The central value.

**Formula:**  
$\text{Mean} = \frac{x_1 + x_2 + ... + x_n}{n}$

**Example:** `(12 + 15 + 14 + 10 + 18) / 5 = 13.8`

```python
data = np.array([12,15,14,10,18])
data.mean()
```

## **1.2 Median (Middle Value)**
**Meaning:** The value at the center when numbers are sorted.

**Formula:**  
Odd count → middle value  
Even count → average of two middle values

**Example:** Sorted → `[10, 12, 14, 15, 18]` → median = 14

```python
np.median(data)
```

## **1.3 Mode (Most Frequent Value)**
**Meaning:** The value that appears the most.

**Example:** `[12, 15, 15, 18]` → mode = 15

```python
from collections import Counter
Counter(data).most_common(1)[0][0]
```

## **1.4 Variance & Standard Deviation**
**Variance Meaning:** How spread out the data is from the mean.

**Formula:**
$\text{Variance} = \frac{1}{n} \sum (x_i - \text{mean})^2$

**Std Dev:** Square root of variance.
$\text{SD} = \sqrt{\text{variance}}$

**Example:** If values are far from mean → variance is large.

```python
np.var(data), np.std(data)
```

# **2. Linear Algebra Basics**

Used in: Machine Learning models, PCA, Neural Networks.

## **2.1 Vectors**
A vector is just a list of numbers.

**Example:** Person features → height, weight → `[170, 72]`

```python
v = np.array([170, 72]); v
```

## **2.2 Dot Product**
**Meaning:** Multiply matching numbers and add.

**Formula:**
$a·b = a_1b_1 + a_2b_2 + ... + a_nb_n$

**Example:** `[1,2,3]·[4,5,6] = 1*4 + 2*5 + 3*6 = 32`

```python
np.dot(np.array([1,2,3]), np.array([4,5,6]))
```

## **2.3 Norm (Vector Length)**
**Formula:**
$||v|| = \sqrt{x_1^2 + x_2^2 + ...}$

**Example:** Vector `[3,4]` → distance = 5

```python
np.linalg.norm(np.array([3,4]))
```

## **2.4 Matrices**
Matrices store data or transformations.

Example: dataset → each row = sample, columns = features

```python
A = np.array([[1,2],[3,4]]); A
```

## **2.5 Matrix Multiplication**
**Formula:** Multiply rows with columns.

A useful in: Linear Regression, Neural Networks

```python
B = np.array([[2,0],[1,3]]); A @ B
```

## **2.6 Eigenvalues / Eigenvectors**
**Meaning:** Special directions of data.

**Used in:** PCA (feature reduction)

**Formula:**
$A v = λ v$

```python
M = np.array([[2,1],[1,2]])
np.linalg.eig(M)
```

# **3. Calculus Basics**

## **3.1 Derivatives**
**Meaning:** Rate of change.

**Example:**
f(x) = x² → derivative = 2x

Used in Gradient Descent!

```python
def f(x): return x**2
(x, f(x), 2*x) for x in range(-2,3)
```

## **3.2 Gradient**
**Meaning:** Derivative for multiple variables.

**Example:** f(x,y)=x²+y² → gradient = [2x, 2y]

```python
def grad(x,y): return np.array([2*x,2*y]); grad(3,4)
```

## **3.3 Gradient Descent (Optimization)**
Goal: Find minimum of a function.

**Update formula:**
$x = x - α * (gradient)$

```python
lr = 0.1
x = 5
for _ in range(10): x = x - lr*(2*x)
x
```

# **4. Probability Basics**

## **4.1 Probability**
Chance of an event. Used in ML models like Naive Bayes.

```python
np.random.choice([0,1], p=[0.3,0.7], size=10)
```

## **4.2 Expected Value**
**Formula:**
$E[X] = Σ (x_i * P(x_i))$

**Example:** 0 with 30%, 1 with 70% → EV = 0*0.3 + 1*0.7 = 0.7

```python
data = np.array([0,1]); probs = np.array([0.3,0.7]); np.sum(data*probs)
```

## **4.3 Covariance & Correlation**
**Covariance formula:**
$cov(X,Y)=\frac{Σ (x_i-mean_x)(y_i-mean_y)}{n-1}$

**Correlation = normalized covariance**

```python
x = np.random.randn(50)
y = 2*x + np.random.randn(50)
np.corrcoef(x,y)
```

# **5. PCA (Principal Component Analysis)**

Used to reduce features.

Steps:
1. Center data
2. Apply SVD
3. First eigenvector = main direction of variance

```python
X = np.vstack([x,y]).T
Xc = X - X.mean(axis=0)
U,s,VT = np.linalg.svd(Xc)
VT[0]
```
