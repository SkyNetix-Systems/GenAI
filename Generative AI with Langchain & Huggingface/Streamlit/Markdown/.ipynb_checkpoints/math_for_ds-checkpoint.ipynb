## 1 — Setup

Import common libraries. All plots use `matplotlib` and appear inline.

```
import numpy as np
import matplotlib.pyplot as plt
import math

np.set_printoptions(precision=4, suppress=True)
print('numpy version:', np.__version__)
```

## 2 — Vectors and basic operations

A vector is an array of numbers. Key operations: addition, scalar multiplication, dot product, and norm.

```
# Examples
v = np.array([1, 2, 3])
w = np.array([4, 5, 6])

print('v + w =', v + w)
print('2 * v =', 2 * v)
print('dot(v,w) =', v.dot(w))
print('||v|| (L2 norm) =', np.linalg.norm(v))
print('||v||^2 =', np.dot(v, v))
```

### Geometric intuition (plotting two vectors in 2D)

```
u = np.array([2, 1])
v2 = np.array([1, 3])

plt.figure()
plt.arrow(0,0,u[0],u[1],head_width=0.15)
plt.arrow(0,0,v2[0],v2[1],head_width=0.15)
plt.xlim(-1,4)
plt.ylim(-1,4)
plt.gca().set_aspect('equal', adjustable='box')
plt.title('Two vectors in 2D')
plt.grid(True)
plt.show()
```

## 3 — Matrices

A matrix is a 2D array. Common operations: multiplication, transpose, inverse (for square invertible matrices).

```
A = np.array([[1,2],[3,4]])
B = np.array([[2,0],[1,3]])

print('A:\n', A)
print('\nB:\n', B)
print('\nA @ B:\n', A @ B)
print('\nA.T:\n', A.T)
print('\ninv(A):\n', np.linalg.inv(A))
```

### Linear system example

Solve Ax = b using the matrix inverse or `np.linalg.solve`.

```
A = np.array([[3,1],[1,2]])
b = np.array([9,8])

# solution x
x1 = np.linalg.inv(A) @ b
x2 = np.linalg.solve(A, b)
print('solution using inverse:', x1)
print('solution using solve:', x2)
```

## 4 — Eigenvalues & Eigenvectors (intuition)

Eigenvectors are directions that the matrix transforms by only scaling (the eigenvalue). They are useful in PCA, stability analysis, and more.

```
M = np.array([[2,1],[1,2]])
vals, vecs = np.linalg.eig(M)
print('eigenvalues =', vals)
print('eigenvectors (columns) =\n', vecs)
```

## 5 — Singular Value Decomposition (SVD)

SVD decomposes a matrix `X = U S V^T`. It's used in PCA, low-rank approximations, and recommendation systems.

```
X = np.array([[3,1,1],[1,3,1]])
U, s, VT = np.linalg.svd(X, full_matrices=False)
print('U =\n', U)
print('\ns =', s)
print('\nVT =\n', VT)
print('\nReconstruct X from U*s*VT:')
Sigma = np.diag(s)
print(U @ Sigma @ VT)
```

## 6 — Calculus: derivatives (brief)

Derivative of a function f(x) measures its rate of change. For multivariable functions we use gradients (vector of partial derivatives).

```
# Derivative demonstration for f(x) = x^2
def f(x):
    return x**2

def numerical_derivative(func, x, h=1e-6):
    return (func(x+h)-func(x-h)) / (2*h)

xs = np.array([-2.0, -1.0, 0.0, 1.0, 2.0])
print('x, f(x), f\'(x) approx')
for x in xs:
    print(x, f(x), numerical_derivative(f, x))
```

## 7 — Gradients & multivariable example

For f(x,y) = x^2 + y^2, the gradient is [2x, 2y]. We'll compute and visualize gradient vectors on a grid.

```
def grad_f(x, y):
    return np.array([2*x, 2*y])

# grid for plotting
xs = np.linspace(-2, 2, 9)
ys = np.linspace(-2, 2, 9)
Xg, Yg = np.meshgrid(xs, ys)
U = 2 * Xg
V = 2 * Yg

plt.figure()
plt.quiver(Xg, Yg, U, V)
plt.title('Gradient field of f(x,y)=x^2+y^2')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

## 8 — Optimization: Gradient Descent for Linear Regression

Simple linear regression `y = w0 + w1*x` fit by minimizing MSE. We'll use gradient descent (manual) so students see the updates.

```
# Simulate simple linear data
np.random.seed(0)
X = 2 * np.random.rand(100)
true_w0 = 1.5
true_w1 = 2.7
Y = true_w0 + true_w1 * X + np.random.randn(100) * 0.5

# Add bias term
X_b = np.c_[np.ones((100,1)), X]

# Gradient descent parameters
lr = 0.1
n_iter = 50
m = 100
theta = np.random.randn(2)

history = []
for i in range(n_iter):
    gradients = (2/m) * X_b.T.dot(X_b.dot(theta) - Y)
    theta = theta - lr * gradients
    mse = np.mean((X_b.dot(theta) - Y)**2)
    history.append(mse)

print('Estimated theta (w0, w1):', theta)
print('True (w0, w1):', [true_w0, true_w1])

plt.figure()
plt.plot(history)
plt.title('MSE during Gradient Descent')
plt.xlabel('Iteration')
plt.ylabel('MSE')
plt.show()

# Scatter + fit line
plt.figure()
plt.scatter(X, Y)
xs = np.array([0,2])
plt.plot(xs, theta[0] + theta[1]*xs)
plt.title('Linear fit by gradient descent')
plt.show()
```

## 9 — Probability basics

Expectation (mean), variance, covariance, and correlation are central. We'll compute them for sample data.

```
data = np.random.normal(loc=5, scale=2, size=1000)
print('sample mean =', data.mean())
print('sample var =', data.var(ddof=1))

# Covariance and correlation between two variables
x = np.random.normal(0,1,500)
y = 2*x + np.random.normal(0,1,500)

cov = np.cov(x, y, ddof=1)
cor = np.corrcoef(x, y)
print('\nCovariance matrix:\n', cov)
print('\nCorrelation matrix:\n', cor)
```

## 10 — PCA demo (using SVD)

PCA finds principal components (directions of maximum variance). We'll create 2D correlated data and show PCA projection.

```
np.random.seed(1)
# correlated data
x = np.random.normal(0,1,200)
y = 0.8 * x + np.random.normal(0,0.6,200)
X = np.vstack([x,y]).T

# center data
X_centered = X - X.mean(axis=0)
U, s, VT = np.linalg.svd(X_centered, full_matrices=False)
pc1 = VT[0]
print('First principal component:', pc1)

# project
proj = X_centered.dot(pcv := pc1)

plt.figure()
plt.scatter(X[:,0], X[:,1])
# plot PC direction line through mean
mean = X.mean(axis=0)
line_x = np.array([-3, 3])
line = np.outer(line_x, pcv[:2])
plt.plot(mean[0] + line[:,0], mean[1] + line[:,1])
plt.title('Data and first principal component (line)')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
