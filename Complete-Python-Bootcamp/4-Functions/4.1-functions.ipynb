```python
## why functions?
num=24
if num%2==0:
    print("the number is even")
else:
    print("the number is odd")
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print("the number is even")
    else:
        print("the number is odd")
```

```python
## Call this function
even_or_odd(24)
```

```python
## Positional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Krish")
```

```python
### Keywords Arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Krish",age="32",country="India")
```

Explanation

**kwargs means the function can take any number of keyword arguments.

Keyword argument = name="Krish", age=25, etc.

Inside the function, kwargs becomes a dictionary where:

Keys → argument names

Values → argument values

.items() lets you loop through each key–value pair in that dictionary.

The print(f"{key}:{value}") line prints each detail in a clean "key:value" format.

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f" Positional arument :{val}")
    
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"Krish",name="Krish",age="32",country="India")
```

🧠 Explanation:

This function accepts two types of inputs:

⭐ 1. Positional Arguments (*args)

These are values passed without names.
Inside the function, they’re collected into a tuple.
The function loops through them and prints each one as a positional argument.

⭐ 2. Keyword Arguments (**kwargs)

These are arguments passed as key=value pairs.
Inside the function, they’re stored in a dictionary.
The function loops through this dictionary and prints each key along with its value.

🎯 What the function does overall

Prints all positional arguments first

Then prints all keyword arguments

Handles any number of both

Super useful when writing flexible, dynamic functions
