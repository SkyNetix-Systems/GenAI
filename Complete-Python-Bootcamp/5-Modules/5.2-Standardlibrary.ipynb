##### Standard Library Overview
Python's Standard Library is a vast collection of modules and packages that come bundled with Python, providing a wide range of functionalities out of the box. Here's an overview of some of the most commonly used modules and packages in the Python Standard Library.

📄 Module

A single Python file (.py)

Contains code: variables, functions, classes, etc.

Think of it like one chapter in a book.

Example:
math.py, utils.py, calculator.py

📦 Package

A folder/directory that contains multiple modules.

Must have an __init__.py file (even empty) to be recognized as a package.

Think of it like a whole book, made of many chapters (modules).

```python
import array
arr=array.array('i',[1,2,3,4])
print(arr)
print(type(arr))
```

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
## Data Serialization
import json
data={'name':'Krish','age':25}

# json to string
json_str=json.dumps(data)
print(json_str)
print(type(json_str))

# string to json
parsed_data=json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## csv

import csv

with open('example.csv',mode='w',newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Akhilesh',32])

with open('example.csv',mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
## datetime
from datetime import datetime,timedelta

now=datetime.now()
print(now)

yesterday=now-timedelta(days=1)

print(yesterday)
```

```python
## time
import time
print(time.time())
time.sleep(2)
print(time.time())
```
