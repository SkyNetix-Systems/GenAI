#### Sets
Sets are a built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference, and symmetric difference.

## 🧠 Python Set discard() Explanation

```python
my_set.discard(11)
print(my_set)
```

### 🔍 Explanation

### 1️⃣ `my_set.discard(11)`
- Tries to remove the element **11** from `my_set`.
- If **11 exists**, it removes it.
- If **11 does NOT exist**, it **does nothing** (no error, no exception).
- This is the key difference between `discard()` and `remove()` —  
  `remove()` throws an error if the element is missing,  
  but `discard()` stays chill 😎.

### 2️⃣ `print(my_set)`
- Prints the updated set after attempting to discard **11**.

### 🎯 Final Result
- If **11 was in the set**, it will be removed.
- If **11 was NOT in the set**, the set stays unchanged.

## 🔢 Set Operations in Python — Explained

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}
```

---

## 🔻 Difference
```python
print(set1.difference(set2))
set2.difference(set1)
```

### 🧠 Explanation
- **Difference** returns elements that exist in the first set but **not** in the second.
- `set1.difference(set2)` → `{1, 2, 3}`
- `set2.difference(set1)` → `{7, 8, 9}`
- It does **not** modify the original sets.

---

## 🔷 Symmetric Difference
```python
set1.symmetric_difference(set2)
```

### 🧠 Explanation
- Returns elements that are **NOT shared** by both sets.
- Basically: all unique elements from both sets **excluding** the intersection.
- Result: `{1, 2, 3, 7, 8, 9}`

---

## 🧪 Set Methods: Subset & Superset

```python
set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5}
```

### 🔹 is subset
```python
print(set1.issubset(set2))
```

### 🧠 Explanation
- Checks whether **all elements of set1** exist inside set2.
- Result: `False`

### 🔹 is superset
```python
print(set1.issuperset(set2))
```

### 🧠 Explanation
- Checks whether set1 contains **all elements of set2**.
- Result: `True`
