# Descriptive Statistics — Mean, Median, Mode, Variance & Standard Deviation

This notebook explains each concept with a clear definition, a simple numeric example, and runnable Python code that computes the measure and visualizes the data. Use it to teach students step-by-step. 🎓📊

---

## Setup
Import libraries and set display options.

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

np.set_printoptions(precision=4, suppress=True)
plt.style.use('default')

print('numpy', np.__version__)
print('pandas', pd.__version__)
```

## Dataset 1 — Small simple dataset
We'll use a small dataset to compute each statistic by hand and with code.

`data = [12, 15, 14, 10, 18, 18, 20]`

```
data = np.array([12, 15, 14, 10, 18, 18, 20])
data
```

## 1 — Mean (Average)
**Definition:** Sum of values divided by count.

**Hand calculation example:**

Mean = (12 + 15 + 14 + 10 + 18 + 18 + 20) / 7 = 107 / 7 ≈ 15.2857

**Compute in Python:**

```
mean_val = data.mean()
mean_val
```

**Plot:** Show mean on the histogram.

```
plt.figure(figsize=(6,3))
plt.hist(data, bins=7, edgecolor='k')
plt.axvline(mean_val, color='r', linestyle='--', label=f'Mean = {mean_val:.2f}')
plt.title('Histogram with Mean')
plt.legend()
plt.show()
```

## 2 — Median (Middle Value)
**Definition:** Middle value when data are sorted. If even count, median is average of two middle values.

**Hand example:** Sorted data = [10,12,14,15,18,18,20] → middle is 15.

**Compute in Python:**

```
median_val = np.median(data)
median_val
```

**Plot:** Show median on the histogram.

```
plt.figure(figsize=(6,3))
plt.hist(data, bins=7, edgecolor='k')
plt.axvline(median_val, color='g', linestyle='-.', label=f'Median = {median_val:.2f}')
plt.title('Histogram with Median')
plt.legend()
plt.show()
```

## 3 — Mode (Most Frequent Value)
**Definition:** Value(s) that appear most frequently.

**Hand example:** Data has 18 twice → mode = 18.

**Compute in Python:**

```
# Using Counter for mode (works well for numeric or categorical data)
counts = Counter(data)
mode_count = counts.most_common()
mode_count

# Extract the top mode(s)
max_freq = mode_count[0][1]
modes = [val for val, freq in mode_count if freq == max_freq]
print('modes =', modes, 'frequency =', max_freq)
```

**Note:** There can be more than one mode (bimodal, multimodal).

## 4 — Variance
**Definition:** Average squared deviation from the mean. Measures spread.

**Population vs Sample:**
- Population variance uses denominator N. `np.var(data, ddof=0)`
- Sample variance uses denominator (N-1). `np.var(data, ddof=1)` — commonly used when data is a sample.

**Hand intuition:** For our data, compute squared differences from the mean, average them (use N-1 for sample).

```
# Sample variance (ddof=1)
variance_sample = np.var(data, ddof=1)
# Population variance (ddof=0)
variance_pop = np.var(data, ddof=0)
variance_sample, variance_pop
```

**Interpretation:** Larger variance → values spread out more. Units are squared (e.g., if data units are meters, variance units are meters²).

## 5 — Standard Deviation
**Definition:** Square root of variance. Expressed in same units as data.

Compute sample standard deviation (ddof=1):

```
std_sample = np.std(data, ddof=1)
std_sample
```

**Plot:** Show mean ± 1 standard deviation on the histogram.

```
plt.figure(figsize=(6,3))
plt.hist(data, bins=7, edgecolor='k')
plt.axvline(mean_val, color='r', linestyle='--', label=f'Mean = {mean_val:.2f}')
plt.axvline(mean_val - std_sample, color='b', linestyle=':', label=f'Mean - 1 SD = {mean_val-std_sample:.2f}')
plt.axvline(mean_val + std_sample, color='b', linestyle=':', label=f'Mean + 1 SD = {mean_val+std_sample:.2f}')
plt.title('Histogram with Mean ± 1 SD')
plt.legend()
plt.show()
```

## 6 — Worked Example: Outlier effect
Compare mean and median after adding an outlier.

```
data2 = np.append(data, 1000)  # add a large outlier
print('original mean, median =', mean_val, median_val)
print('with outlier mean, median =', np.mean(data2), np.median(data2))

plt.figure(figsize=(8,3))
plt.subplot(1,2,1)
plt.hist(data, bins=7, edgecolor='k')
plt.title('Original')
plt.subplot(1,2,2)
plt.hist(data2, bins=7, edgecolor='k')
plt.title('With outlier (1000)')
plt.tight_layout()
plt.show()
```

## 7 — Summary Table (computed)

```
summary = pd.DataFrame({
    'Statistic': ['Count','Mean','Median','Mode(s)','Sample Variance (ddof=1)','Population Variance (ddof=0)','Sample Std Dev'],
    'Value': [len(data), round(mean_val,4), round(median_val,4), str(modes), round(variance_sample,4), round(variance_pop,4), round(std_sample,4)]
})
summary
```
