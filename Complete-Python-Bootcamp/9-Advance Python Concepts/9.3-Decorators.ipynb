#### Decorators
Decorators are a powerful and flexible feature in Python that allows you to modify the behavior of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. This lesson covers the basics of decorators, including how to create and use them.

It just means:
You wrote a function, and you’re calling it (like making a “copy” of whatever it returns).

Not copying memory or cloning — just invoking the function to get its output.

⭐ First — What is a Closure?

A closure happens when:

You have a function inside another function

The inner function uses a variable from the outer function

You return the inner function itself

That inner function remembers the outer variable even after the outer function finishes

So closure = inner function + remembered data 🔥

````python
def main_welcome(func):
   
    def sub_welcome_method():
        print("Welcome to the advance python course")
        func("Welcome everyone to this tutorial")
        print("Please learn these concepts properly")
    return sub_welcome_method()
````

````python
main_welcome(print)
````

````python
def main_welcome(func,lst):
   
    def sub_welcome_method():
        print("Welcome to the advance python course")
        print(func(lst))
        print("Please learn these concepts properly")
    return sub_welcome_method()
````

````python
main_welcome(len,[1,2,3,4,5])
````

### 🎯 What is a Decorator?

A **decorator** is a function that:
1. Accepts another function  
2. Wraps extra behavior around it  
3. Returns a NEW enhanced function  

Decorators allow you to:
- Add features  
- Log activity  
- Add authentication  
- Time execution  
- Modify behavior  
**Without changing the original function's code**

---

### 🧱 Basic Decorator Structure

```python
def decorator_name(func):
    
    def wrapper():
        # Code before
        func()      # Original function call
        # Code after
    
    return wrapper


````python
## Decorators WWith arguments
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
````

````python
@repeat(3)
def say_hello():
    print("Hello")
````

````python
say_hello()
````
