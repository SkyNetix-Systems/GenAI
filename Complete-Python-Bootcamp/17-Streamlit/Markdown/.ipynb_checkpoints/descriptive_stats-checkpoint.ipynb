# Mean, Median, Mode, Variance & Standard Deviation — Simple Notebook

This notebook teaches the most essential descriptive statistics used in Data Science. Each concept includes a short explanation + Python example. 📊

## 1 — Setup

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print('numpy:', np.__version__)
print('pandas:', pd.__version__)
```

## 2 — Sample Data
We'll work with a small sample dataset.

```
data = np.array([12, 15, 14, 10, 18, 18, 20, 22, 15, 14, 15])
data
```

## 3 — Mean
The **mean** is the average of all values.

```
mean_val = np.mean(data)
mean_val
```

## 4 — Median
The **median** is the middle value when data is sorted.

```
median_val = np.median(data)
median_val
```

## 5 — Mode
The **mode** is the value that appears most frequently.

```
from collections import Counter
counts = Counter(data)
mode_val = counts.most_common(1)[0][0]
mode_val
```

## 6 — Variance
Variance measures how spread out the numbers are.

```
variance_val = np.var(data, ddof=1)  # sample variance
variance_val
```

## 7 — Standard Deviation
The standard deviation is the square root of variance.

```
std_val = np.std(data, ddof=1)
std_val
```

## 8 — Visualizing the Distribution
Histogram helps understand spread.

```
plt.figure(figsize=(6,4))
plt.hist(data, bins=6)
plt.title('Data Histogram')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

## 9 — Summary Table

```
summary = pd.DataFrame({
    'Statistic': ['Mean', 'Median', 'Mode', 'Variance', 'Std Dev'],
    'Value': [mean_val, median_val, mode_val, variance_val, std_val]
})
summary
```
