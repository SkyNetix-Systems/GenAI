# Normal vs Uniform Distribution — Super Simple Notebook

This notebook explains both distributions in the **simplest possible way**, with tiny examples and easy visuals.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Uniform Distribution (Simplest Explanation)
- Every value in a range has **equal chance**.
- Example: picking a random number between 0 and 1.

```python
u = np.random.uniform(0, 1, 10000)
plt.hist(u, bins=20, edgecolor='black')
plt.title('Uniform(0,1) Distribution')
plt.show()
```

**What you observe:** The bars look almost the same height → all values are equally likely.

## 2. Normal Distribution (Simplest Explanation)
- Values cluster around the **mean**.
- Shape looks like a **bell curve**.

```python
n = np.random.normal(0, 1, 10000)
plt.hist(n, bins=30, edgecolor='black')
plt.title('Normal(0,1) Distribution')
plt.show()
```

**What you observe:** Middle bars are tallest (values near 0 appear more), and sides taper off.

## 3. Quick Visual Comparison
Uniform = flat-ish.
Normal = bell-shaped.

```python
fig, ax = plt.subplots(1,2, figsize=(10,4))

ax[0].hist(u, bins=20, edgecolor='black')
ax[0].set_title('Uniform')

ax[1].hist(n, bins=30, edgecolor='black')
ax[1].set_title('Normal')

plt.show()
```

## Summary
- **Uniform:** all values equally likely.
- **Normal:** values cluster around mean, forming a bell curve.
