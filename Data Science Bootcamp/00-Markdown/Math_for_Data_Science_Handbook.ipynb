## 🎲 Probability

### What is Probability?
Probability measures how likely an event is to occur. Values range from **0 to 1**.

### Key Concepts
- **Sample space (Ω)** — set of all possible outcomes  
- **Event (A)** — subset of outcomes  
- **Probability of event A:**  
$$P(A) = \frac{|A|}{|\Omega|}$$

### Conditional Probability
$$P(A|B) = \frac{P(A∩B)}{P(B)}$$

### Bayes’ Theorem
$$P(A|B)=\frac{P(B|A)P(A)}{P(B)}$$

Used in Naive Bayes, Bayesian inference, anomaly detection.
