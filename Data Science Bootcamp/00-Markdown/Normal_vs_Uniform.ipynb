# Normal vs Uniform Distribution — Simple Explanation

This notebook demonstrates the **Normal (Gaussian)** and **Uniform** distributions with simple explanations, sampling examples, summary statistics, and plots.  
Run the cells sequentially.


## 1 — Quick intuitive descriptions

- **Uniform distribution**: every value in a range is equally likely. Think of a fair spinner that lands anywhere between 0 and 1 with equal chance.  
- **Normal distribution**: values cluster around a central mean in a bell-shaped curve. Many natural phenomena (heights, measurement noise) are approximately normal.


### Imports and helper functions

```python
import numpy as np
import matplotlib.pyplot as plt
from math import sqrt, pi, exp

def print_stats(name, samples):
    print(f"{name} — n={len(samples)}")
    print("  mean:", samples.mean())
    print("  variance:", samples.var())
    print("  std dev:", samples.std())
    print()
```

## 2 — Uniform Distribution (example)
Sample from Uniform(0, 1), show histogram and statistics.

```python
rng = np.random.default_rng(0)
u_samples = rng.uniform(low=0.0, high=1.0, size=10000)
print_stats("Uniform(0,1) samples", u_samples)

plt.figure(figsize=(7,4))
plt.hist(u_samples, bins=30, edgecolor='black')
plt.title('Histogram of Uniform(0,1) samples')
plt.xlabel('value'); plt.ylabel('frequency')
plt.show()
```

**Key points (Uniform):**  
- Every value in [0, 1) is roughly equally likely.  
- The theoretical mean is 0.5 and variance is 1/12 ≈ 0.0833.

## 3 — Normal Distribution (example)
Sample from Normal(mean=0, std=1), show histogram and the theoretical PDF overlay.

```python
rng = np.random.default_rng(1)
n_samples = rng.normal(loc=0.0, scale=1.0, size=10000)
print_stats("Normal(0,1) samples", n_samples)

# histogram + theoretical PDF overlay
plt.figure(figsize=(7,4))
count, bins, _ = plt.hist(n_samples, bins=40, density=True, edgecolor='black')

# theoretical normal pdf
bin_centers = 0.5*(bins[1:] + bins[:-1])
def normal_pdf(x, mu=0.0, sigma=1.0):
    return (1.0 / (sqrt(2*pi)*sigma)) * np.exp(-0.5*((x-mu)/sigma)**2)

pdf_vals = normal_pdf(bin_centers)
plt.plot(bin_centers, pdf_vals, linewidth=2)
plt.title('Histogram of Normal(0,1) samples with PDF overlay')
plt.xlabel('value'); plt.ylabel('density')
plt.show()
```

**Key points (Normal):**  
- The familiar bell curve centered at the mean (0 here).  
- For Normal(0,1), theoretical mean = 0, variance = 1.

## 4 — Comparing sampling behavior & CLT demo
We will draw many small-sample means from Uniform and Normal distributions and see their distributions (Central Limit Theorem).

```python
# Draw many sample means for CLT demo
rng = np.random.default_rng(2)
n_repeats = 2000
sample_size = 30

means_uniform = [rng.uniform(0,1,sample_size).mean() for _ in range(n_repeats)]
means_normal = [rng.normal(0,1,sample_size).mean() for _ in range(n_repeats)]

plt.figure(figsize=(7,4))
plt.hist(means_uniform, bins=30, edgecolor='black')
plt.title('Distribution of sample means (Uniform(0,1), sample size=30)')
plt.xlabel('sample mean'); plt.ylabel('frequency')
plt.show()

plt.figure(figsize=(7,4))
plt.hist(means_normal, bins=30, edgecolor='black')
plt.title('Distribution of sample means (Normal(0,1), sample size=30)')
plt.xlabel('sample mean'); plt.ylabel('frequency')
plt.show()

print_stats("Means of Uniform samples", np.array(means_uniform))
print_stats("Means of Normal samples", np.array(means_normal))
```

**What to notice:**  
- Regardless of the original distribution (uniform or normal), the distribution of sample *means* tends to a normal shape when the sample size is large enough — that's the Central Limit Theorem (CLT).  
- The CLT is why normal approximations show up a lot in statistics and ML.
