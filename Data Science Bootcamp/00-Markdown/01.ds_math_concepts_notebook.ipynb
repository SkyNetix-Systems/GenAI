## Mean (Average)

The **mean** of a dataset is the sum of values divided by the count:

$$\text{Mean} = \mu = \frac{1}{n}\sum_{i=1}^n x_i$$

Example and computation with NumPy:


```python
import numpy as np
data = np.array([10, 20, 30, 40, 50])
print('data =', data)
mean_val = np.mean(data)
print('Mean =', mean_val)
```

## Median

The **median** is the middle value when the data are sorted. For even number of observations it's the average of the two middle values.

Example:


```python
data_odd = np.array([10, 20, 30, 40, 50])
data_even = np.array([10, 20, 30, 40])
print('Median (odd) =', np.median(data_odd))
print('Median (even) =', np.median(data_even))
```

## Variance and Standard Deviation

Variance measures the average squared deviation from the mean, and standard deviation is the square root of variance.

$$\text{Variance} = \sigma^2 = \frac{1}{n}\sum_{i=1}^n (x_i - \mu)^2$$
$$\text{Standard Deviation} = \sigma = \sqrt{\sigma^2}$$

Population (default NumPy) formulas are shown above. Example below:


```python
data = np.array([10, 12, 14, 16, 18])
print('Data =', data)
print('Variance =', np.var(data))
print('Std Dev =', np.std(data))
```

## Normal Distribution (Gaussian)

A normal distribution is a continuous probability distribution defined by mean $\mu$ and standard deviation $\sigma$.

Probability density function (PDF):
$$f(x)=\frac{1}{\sigma\sqrt{2\pi}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

Properties:
- Symmetric around the mean
- Mean = Median = Mode
- Controlled by $\mu$ (center) and $\sigma$ (spread)

Let's draw samples and show a histogram.


```python
import matplotlib.pyplot as plt
np.random.seed(101)
normal_data = np.random.normal(loc=0, scale=1, size=1000)
plt.figure(figsize=(8,4))
plt.hist(normal_data, bins=30)
plt.title('Histogram: Normal Distribution (mu=0, sigma=1)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

## Uniform Distribution

A continuous uniform distribution gives equal probability to all values in the interval $[a, b]$.

PDF for continuous uniform on [a,b]:
$$f(x)=\frac{1}{b-a}\quad\text{for }a\le x\le b$$

Let's draw samples and show a histogram.


```python
uniform_data = np.random.uniform(low=0, high=1, size=1000)
plt.figure(figsize=(8,4))
plt.hist(uniform_data, bins=30)
plt.title('Histogram: Uniform Distribution (0,1)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

## Basic Probability Concepts

Probability is a number between 0 and 1. For an event A:
$$P(A)=\frac{\text{favorable outcomes}}{\text{total outcomes}}$$

Example: rolling a fair 6-sided die, probability of getting a 4 is $1/6$.


```python
# Practical example: compute stats for a synthetic dataset
np.random.seed(42)
data = np.random.normal(loc=50, scale=5, size=200)
print('Mean:', np.mean(data))
print('Median:', np.median(data))
print('Variance:', np.var(data))
print('Std Dev:', np.std(data))

plt.figure(figsize=(8,4))
plt.hist(data, bins=25)
plt.title('Histogram: synthetic normal data (loc=50, scale=5)')
plt.show()
```
