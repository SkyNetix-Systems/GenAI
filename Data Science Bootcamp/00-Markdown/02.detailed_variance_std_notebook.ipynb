## Variance & Standard Deviation (DETAILED EXPLANATION)

Variance and Standard Deviation are two of the most important concepts in statistics and data science.

### 🧠 Why do we need them?
They measure **spread** — how far the data points are from the mean.

---
## 📌 1. Variance
Variance measures **how spread out** the data is. It computes the average of squared differences from the mean.

### 👉 Formula (population variance):
$$\sigma^2 = \frac{1}{N} \sum_{i=1}^{N} (x_i - \mu)^2$$

Where:
- $x_i$ = each data point
- $\mu$ = mean
- $N$ = total number of values

### 🔍 Why square the differences?
- To avoid positive/negative canceling out
- To give more weight to larger deviations

### 🤓 Intuition
- Variance is **big** → points are far apart
- Variance is **small** → points are close to the mean

---
## 📌 2. Standard Deviation (Std Dev)
Standard deviation is simply the **square root of variance**:

$$\sigma = \sqrt{\sigma^2}$$

### Why take the square root?
Because variance is in **squared units**.
- If data is in meters → variance is in square meters
- Std deviation brings it back to meters

### Std Dev tells you:
- How much data typically **deviates from the mean**
- Helps understand spread in the **original units**

### 📌 Rule of Thumb (Normal Distribution):
- 68% of data lies within **±1 standard deviation**
- 95% within **±2 standard deviations**
- 99.7% within **±3 standard deviations**

---
## 📊 Example Calculation (Step-by-Step)

---

### **Step 1: Mean**

Sum = 10 + 12 + 14 + 16 + 18 = 70  
Mean = 70 / 5 = **14**

---

### **Step 2: Squared Deviations**

- (10 − 14)² = 16  
- (12 − 14)² = 4  
- (14 − 14)² = 0  
- (16 − 14)² = 4  
- (18 − 14)² = 16  

Total squared deviations = 16 + 4 + 0 + 4 + 16 = **40**

---

### **Step 3: Variance**

Variance = Total Squared Deviations / Number of Values  
Variance = 40 / 5 = **8**

---

### **Step 4: Standard Deviation**

Std Dev = √8 ≈ **2.828**

---

### **Interpretation**

The numbers are evenly spaced, so the spread is super consistent — each value sits about **2.828 units** away from the mean.

---
### Code Example


```python
import numpy as np
data = np.array([10, 12, 14, 16, 18])
print('Data:', data)
print('Mean:', np.mean(data))
print('Variance:', np.var(data))
print('Standard Deviation:', np.std(data))
```

## Normal Distribution
A bell-shaped symmetric distribution defined by mean (μ) and std deviation (σ).


```python
import matplotlib.pyplot as plt
np.random.seed(101)
normal_data = np.random.normal(0, 1, 1000)
plt.hist(normal_data, bins=30)
plt.title('Normal Distribution')
plt.show()
```

## Uniform Distribution
All values within a range have equal probability.


```python
uniform_data = np.random.uniform(0, 1, 1000)
plt.hist(uniform_data, bins=30)
plt.title('Uniform Distribution')
plt.show()
```
