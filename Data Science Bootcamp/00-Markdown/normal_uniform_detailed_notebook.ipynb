```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
plt.style.use('seaborn-v0_8')
np.random.seed(42)
```

## Normal Distribution (Gaussian)

**Definition:** A continuous probability distribution characterized by mean $\mu$ and standard deviation $\sigma$.

**PDF (probability density function):**
$$f(x)=\frac{1}{\sigma\sqrt{2\pi}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

**CDF (cumulative distribution function):**
$$F(x)=\frac{1}{2}\left[1+\operatorname{erf}\left(\frac{x-\mu}{\sigma\sqrt{2}}\right)\right]$$

Properties:
- Mean = $\mu$
- Variance = $\sigma^2$
- Symmetric around mean
- 68-95-99.7 rule for ±1, ±2, ±3 std dev


```python
# Sample and compare empirical histogram with theoretical PDF
mu, sigma = 0, 1
samples = np.random.normal(loc=mu, scale=sigma, size=5000)
count, bins, _ = plt.hist(samples, bins=50, density=True, alpha=0.6, label='Empirical')
x = np.linspace(bins[0], bins[-1], 200)
plt.plot(x, stats.norm.pdf(x, mu, sigma), 'r--', lw=2, label='Theoretical PDF')
plt.title('Normal Distribution: empirical vs theoretical')
plt.legend()
plt.show()
```

```python
from statsmodels.distributions.empirical_distribution import ECDF
ecdf = ECDF(samples)
x = np.linspace(samples.min(), samples.max(), 200)
plt.plot(x, ecdf(x), label='Empirical CDF')
plt.plot(x, stats.norm.cdf(x, mu, sigma), 'r--', label='Theoretical CDF')
plt.title('Normal Distribution CDF: empirical vs theoretical')
plt.legend()
plt.show()
```

### Central Limit Theorem (CLT) Demonstration

The CLT states that the distribution of the sample mean of independent random variables (with finite mean and variance) approaches a normal distribution as sample size increases.
We'll demonstrate by sampling from a uniform distribution and plotting the distribution of sample means for increasing sample sizes.

```python
def clt_demo(dist_func, dist_args, sample_size, n_trials=5000):
    means = [np.mean(dist_func(*dist_args, size=sample_size)) for _ in range(n_trials)]
    return np.array(means)

sample_sizes = [1, 2, 5, 10, 30]
plt.figure(figsize=(12,8))
for i, s in enumerate(sample_sizes, 1):
    means = clt_demo(np.random.uniform, (0,1), sample_size=s, n_trials=2000)
    plt.subplot(3,2,i)
    plt.hist(means, bins=30, density=True)
    plt.title(f'sample size = {s}')
plt.tight_layout()
plt.show()
```

### Parameter Estimation for Normal Distribution
Given a sample, estimate mean and standard deviation (MLE for normal):
$$\hat{\mu} = \frac{1}{n}\sum x_i, \quad \hat{\sigma}^2 = \frac{1}{n}\sum (x_i-\hat{\mu})^2$$


```python
sample = np.random.normal(loc=5, scale=2, size=1000)
est_mu = np.mean(sample)
est_sigma = np.std(sample)
print('Estimated mu:', est_mu)
print('Estimated sigma:', est_sigma)
```

## Uniform Distribution (Continuous)

Definition: Uniform distribution on [a, b] assigns equal probability density to every value in the interval.

PDF:
$$f(x)=\frac{1}{b-a} \quad \text{for } a\le x\le b$$

Mean and variance:
$$\text{Mean} = \frac{a+b}{2}$$
$$\text{Variance} = \frac{(b-a)^2}{12}$$


```python
a, b = 0, 1
u_samples = np.random.uniform(low=a, high=b, size=5000)
plt.figure(figsize=(8,4))
plt.hist(u_samples, bins=50, density=True, alpha=0.7)
x = np.linspace(a, b, 200)
plt.plot(x, [1/(b-a)]*len(x), 'r--', lw=2, label='Theoretical PDF')
plt.title('Uniform Distribution: empirical vs theoretical')
plt.legend()
plt.show()
```

```python
ecdf_u = ECDF(u_samples)
x = np.linspace(a, b, 200)
plt.plot(x, ecdf_u(x), label='Empirical CDF')
plt.plot(x, stats.uniform.cdf(x, loc=a, scale=b-a), 'r--', label='Theoretical CDF')
plt.title('Uniform CDF: empirical vs theoretical')
plt.legend()
plt.show()
```

## Sampling Convergence & Law of Large Numbers
As sample size increases, sample mean converges to the theoretical mean (LLN). We'll show the running mean for uniform samples.

```python
samples = np.random.uniform(0,1,1000)
running_mean = np.cumsum(samples)/np.arange(1, len(samples)+1)
plt.figure(figsize=(10,4))
plt.plot(running_mean, label='Running mean')
plt.hlines((a+b)/2, 0, len(samples), colors='r', linestyles='dashed', label='Theoretical mean')
plt.title('Running mean (Law of Large Numbers)')
plt.legend()
plt.show()
```

## Use-cases & Intuition
- **Normal**: modeling measurement noise, many natural/aggregate phenomena, assumptions in statistical tests and linear models.
- **Uniform**: random sampling in a bounded interval, simulation inputs when all outcomes equally likely.

## Summary
- Normal is parameterized by mean and sd (mu, sigma) and appears widely due to CLT.
- Uniform gives equal probability across an interval and has simple closed-form mean/variance.
