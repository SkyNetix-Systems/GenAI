## 1 — Headings

Use `#` for headings. More `#` means smaller heading.

Examples:

```markdown
# H1
## H2
### H3
```

Rendered example below:

````python
from IPython.display import Markdown, display

md = '''# H1\n## H2\n### H3\n'''
display(Markdown(md))
````

## 2 — Emphasis (Bold & Italic)

- *Italic* → wrap text with single asterisks or underscores: `*text*` or `_text_`.
- **Bold** → double asterisks or underscores: `**text**` or `__text__`.
- ***Bold + Italic*** → `***text***`.

Example:

````python
display(Markdown('This is *italic*, this is **bold**, and this is ***both***.'))
````

## 3 — Lists

### Unordered lists
Use `-`, `*`, or `+`:

```markdown
- Item 1
- Item 2
  - Subitem
```

### Ordered lists
Use numbers followed by a period:

```markdown
1. First
2. Second
   1. Nested first
```

````python
md = '''- Item 1
- Item 2
  - Subitem

1. First
2. Second
   1. Nested first
'''
display(Markdown(md))
````

## 4 — Links & Images

- Link: `[label](http://example.com)`
- Image: `![alt text](image_url)`

Example:

```markdown
[OpenAI](https://openai.com)

![Sample image](https://via.placeholder.com/150)
```

````python
md = '''[OpenAI](https://openai.com)\n\n![Placeholder image](https://via.placeholder.com/150)'''
display(Markdown(md))
````

## 5 — Code

Inline code uses backticks: `` `inline` ``.

Code blocks use triple backticks, optionally with a language for syntax highlighting:

```markdown
```python
print('hello')
```
```

````python
display(Markdown('Inline code example: `print("hi")`\n\n```python\nprint("Hello, Markdown!")\n```'))
````

## 6 — Tables

Basic table syntax:

```markdown
| Name | Role |
|---|---|
| Alice | Developer |
| Bob | Designer |
```

Note: Tables are best supported on GitHub and in many Markdown renderers, but some flavors differ.

````python
display(Markdown('| Name | Role |\n|---|---|\n| Alice | Developer |\n| Bob | Designer |'))
````

## 7 — Blockquotes & Horizontal Rules

- Blockquote: `> quoted text`
- Horizontal rule: `---` or `***`

Example:

````python
display(Markdown('> This is a blockquote.\n\n---'))
````

## 8 — Task Lists (GitHub Flavored Markdown)

- [ ] Task not done
- [x] Task done

Useful for checklists in issues and READMEs.

````python
display(Markdown('- [ ] Write the docs\n- [x] Set up repo'))
````

## 9 — Advanced Tips

- Use HTML for fine-grained control when Markdown falls short (e.g., footnotes, styling), but prefer Markdown for portability.
- Keep lines short for better diffs in version control.
- Use fenced code blocks with language tags for syntax highlighting.
- For long documents, organize with a table of contents (manual or generated tools).
