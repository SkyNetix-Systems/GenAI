Everything is minimal, intuitive, and uses NumPy only.

## **Setup**

```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## **1. Vectors**
A vector is just a list of numbers.

```python
v = np.array([1, 2, 3])
w = np.array([4, -1, 2])

print('v =', v)
print('w =', w)

print('v + w =', v + w)
print('2 * v =', 2 * v)
print('Dot product =', np.dot(v, w))
print('||v|| =', np.linalg.norm(v))
```

### **Plotting 2D vectors**

```python
v2 = np.array([2, 1])
w2 = np.array([1, 3])

plt.figure(figsize=(4,4))
plt.axhline(0, color='gray')
plt.axvline(0, color='gray')
plt.quiver(0,0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, label='v2')
plt.quiver(0,0, w2[0], w2[1], angles='xy', scale_units='xy', scale=1, label='w2')
plt.xlim(-1,4)
plt.ylim(-1,4)
plt.gca().set_aspect('equal')
plt.legend()
plt.show()
```

## **2. Matrices**

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 2]])

print('A =\n', A)
print('B =\n', B)
print('Aᵀ =\n', A.T)
print('A × B =\n', A @ B)
```

## **3. Solving Linear Systems**
Solve Ax = b

```python
A = np.array([[2, 1], [5, 3]])
b = np.array([4, 13])

x = np.linalg.solve(A, b)
print('Solution x =', x)
print('Check A @ x =', A @ x)
```

## **4. Determinant & Inverse**

```python
A = np.array([[1, 2], [3, 4]])
print('det(A) =', np.linalg.det(A))
print('A⁻¹ =\n', np.linalg.inv(A))
```

## **5. Eigenvalues & Eigenvectors**

```python
M = np.array([[2, 1], [1, 2]])
vals, vecs = np.linalg.eig(M)

print('Eigenvalues:', vals)
print('Eigenvectors:\n', vecs)
```

## **6. Singular Value Decomposition (SVD)**

```python
X = np.array([[3, 1], [2, 2], [1, 3]], dtype=float)
U, s, VT = np.linalg.svd(X, full_matrices=False)

print('U =\n', U)
print('Singular values =', s)
print('Vᵀ =\n', VT)

print('\nReconstruction:\n', U @ np.diag(s) @ VT)
```
