# ⭐ 1. Descriptive Statistics — Understanding Numbers
This helps us quickly understand a bunch of numbers (like student marks or sales data).

## 🔸 Mean (Average)
**Meaning:** Add all numbers and divide by how many numbers are there.

**Formula:**
```
mean = sum of numbers / how many numbers
```
**Example:**
Numbers = 10, 20, 30 → mean = (10+20+30)/3 = 20

````
data = np.array([10,20,30])
data.mean()
````

## 🔸 Median (Middle Number)
**Meaning:** The value in the middle when numbers are sorted.

Example:
Numbers = 5, 9, 100 → **median = 9** (middle value)


````
np.median([5,9,100])
````

## 🔸 Mode (Most Common Number)
**Meaning:** The value that appears the most.

Example:
Numbers = [2, 3, 3, 7] → **mode = 3**

````
Counter([2,3,3,7]).most_common(1)[0][0]
````

## 🔸 Variance & Standard Deviation (How Spread Out Numbers Are)
- If numbers are close together → **low** variance
- If numbers are very different → **high** variance

````
nums = np.array([10,12,14,16])
np.var(nums), np.std(nums)
````

## 📊 Let’s see the data on a chart

````
marks = np.array([12, 15, 14, 10, 18, 18, 20])
plt.hist(marks, bins='auto')
plt.title('Marks Distribution')
plt.show()
````

# ⭐ 2. Linear Algebra — Understanding Data as Numbers
Machines see data as **vectors** (lists of numbers).

## 🔸 Vectors (Just a List of Numbers)
Example:
A person → [height, weight, age] → [170, 70, 25]

````
person = np.array([170,70,25]); person
````

## 🔸 Dot Product (A Way to Compare Two Vectors)
**Meaning:** Multiply matching numbers and add.

Example: [1,2] and [3,4]
→ 1*3 + 2*4 = 11

````
np.dot([1,2],[3,4])
````

## 🔸 Norm (Size or Length of Vector)
Example: [3,4] → size = 5 (famous 3–4–5 triangle)

````
np.linalg.norm([3,4])
````

# ⭐ 3. Calculus — Understanding Change
Don’t worry! We explain it super simply.

## 🔸 Derivative (How Fast Something Changes)
Example:
If speed is changing, derivative tells how fast the speed is changing.

Let’s take a simple function:
```
f(x) = x²
```
Its change (derivative) is:
```
f’(x) = 2x
```

````
def f(x): return x**2
# approximate derivative
[(x, (f(x+0.001)-f(x))/0.001) for x in [0,1,2,3]]
````

## 🔸 Gradient Descent (How ML Learns)
Machine learning tries to **reduce error** step by step.
Gradient Descent = "keep moving downhill until you reach the lowest point".

# ⭐ 4. Probability — Understanding Uncertainty

## 🔸 Probability Basics
Chance of something happening.
Example:
Coin:
- Heads = 0.5
- Tails = 0.5

## 🔸 Expected Value (Average Outcome)
Example:
Win ₹10 with 70% chance → Expected = 10 * 0.7 = 7

````
import numpy as np
np.mean(np.random.choice([0,10], p=[0.3,0.7], size=1000))
````

# ⭐ 5. Covariance, Correlation & PCA — Relationships & Patterns

## 🔸 Correlation (How Two Things Move Together)
Example:
Study more → score more (positive correlation)
Eat more junk → weight increases (positive)
Exercise more → weight decreases (negative)

````
np.random.seed(0)
x = np.random.randn(100)
y = 2*x + np.random.randn(100)*0.2
np.corrcoef(x,y)
````

## 🔸 PCA (Reduce Features Simply)
PCA finds the **main direction** in the data.
Useful when data has many columns/features.

````
X = np.vstack([x,y]).T
Xc = X - X.mean(axis=0)
U,s,VT = np.linalg.svd(Xc)
VT[0]  # direction
````
